### Predict Insurance Charges for New Input

To predict charges for a new individual, we need to gather their information and preprocess it in the same way the training data was handled. This includes converting categorical features to numerical representations and scaling numerical features.

```python
import pandas as pd
import numpy as np

# Get user input for features
def get_user_input():
    print("Please enter the following details for the prediction:")
    age = int(input("Age: "))
    sex = input("Sex (male/female): ").lower()
    bmi = float(input("BMI: "))
    children = int(input("Number of children: "))
    # Explicitly guide user for 'yes' or 'no' for smoker input
    smoker = input("Smoker (yes/no): ").lower()
    region = input("Region (northeast/northwest/southeast/southwest): ").lower()

    # Create a dictionary from the inputs
    user_data = {
        'age': age,
        'sex': sex,
        'bmi': bmi,
        'children': children,
        'smoker': smoker,
        'region': region
    }
    return pd.DataFrame([user_data])

# Get the input from the user
new_data_raw = get_user_input()

# Display raw input for verification
print("\nUser Input:")
display(new_data_raw)
```

### Preprocessing User Input

We will apply the same transformations as done during the training phase:
1. Convert 'sex' and 'smoker' to numerical (0/1).
2. Create 'bmi_category' and then one-hot encode it.
3. One-hot encode the 'region' feature.
4. Scale the numerical features ('age', 'bmi', 'children') using the previously fitted `StandardScaler`.

```python
# Make a copy to avoid modifying the original input dataframe
processed_data = new_data_raw.copy()

# 1. Convert 'sex' and 'smoker' to numerical (0/1)
processed_data['IsFemale'] = processed_data['sex'].map({'male': 0, 'female': 1})
# Handle 'y'/'n' for smoker more robustly
processed_data['IsSmoker'] = processed_data['smoker'].map({'yes': 1, 'no': 0, 'y': 1, 'n': 0})

# Drop original 'sex' and 'smoker' columns
processed_data = processed_data.drop(columns=['sex', 'smoker'])

# 2. Create 'bmi_category' and then one-hot encode it
processed_data['bmi_category'] = pd.cut(
    processed_data['bmi'],
    bins=[0, 18.5, 24.9, 29.9, np.inf],
    labels=['underweight', 'healthy', 'overweight', 'obese']
)

# One-hot encode bmi_category
processed_data = pd.get_dummies(processed_data, columns=['bmi_category'], prefix='bmi_category')

# Ensure all bmi_category dummy columns are present, fill with 0 if not
# (Important for consistency with training data)
expected_bmi_categories = ['bmi_category_underweight', 'bmi_category_healthy', 'bmi_category_overweight', 'bmi_category_obese']
for col in expected_bmi_categories:
    if col not in processed_data.columns:
        processed_data[col] = 0

# 3. One-hot encode the 'region' feature
processed_data = pd.get_dummies(processed_data, columns=['region'], prefix='region')

# Ensure all region dummy columns are present, fill with 0 if not
# (Important for consistency with training data)
expected_regions = ['region_northeast', 'region_northwest', 'region_southeast', 'region_southwest']
for col in expected_regions:
    if col not in processed_data.columns:
        processed_data[col] = 0

# 4. Scale numerical features ('age', 'bmi', 'children')
# The 'scaler' object was fitted in a previous cell (mH7eiVjM9y4M)
# We need to ensure these columns exist and then apply scaling.

# Select numerical columns for scaling
numerical_cols_to_scale = ['age', 'bmi', 'children']
processed_data[numerical_cols_to_scale] = scaler.transform(processed_data[numerical_cols_to_scale])

# Reorder columns to match X_train, filling missing with 0 (for categorical dummies)
# X_train.columns was found to be in `final_features` variable from the kernel state.

final_features = ['age', 'IsFemale', 'bmi', 'children', 'IsSmoker',
                  'region_northeast', 'region_northwest', 'region_southeast', 'region_southwest',
                  'bmi_category_underweight', 'bmi_category_healthy', 'bmi_category_overweight', 'bmi_category_obese']

# Create the final input dataframe with the correct columns and order
# Fill any missing columns (e.g., if a new region/bmi_category was not in the input) with 0
X_new_input = pd.DataFrame(columns=final_features)
for col in final_features:
    if col in processed_data.columns:
        X_new_input[col] = processed_data[col]
    else:
        X_new_input[col] = 0

# Convert boolean columns to integer 0/1 to match the training data format
for col in X_new_input.columns:
    if X_new_input[col].dtype == bool:
        X_new_input[col] = X_new_input[col].astype(int)

print("\nProcessed Input for Prediction:")
display(X_new_input)
```

### Making the Prediction

```python
# Use the trained model to predict the charges
predicted_charge = model.predict(X_new_input)

print(f"\nPredicted Insurance Charge: ${predicted_charge[0]:,.2f}")
```
